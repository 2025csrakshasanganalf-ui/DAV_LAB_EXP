```python
import numpy as np
import pandas as pd
print("numpy version",np.__version__)
print("pandas version",pd.__version__)
```

```python
marks=np.array([78,85,92,67,88])
print("Marks",marks)
print("Type",type(marks))
```

```python
print("Number of dimensions:",marks.ndim)
print("Shape:",marks.shape)
print("Number of elements:",marks.size)
```

```python
marks=np.array([78,85,92,67,88])
print("Original:",marks)
print("Add 5",marks+5)
print("square ",marks**2)
print("Multiply 2",marks*2)
print("Divide 2",marks/2)
```

```python
print("Sum", np.sum(marks))
print("Min", np.min(marks))
print("Max",np.max(marks))
print("Mean",np.mean(marks))
print("Median",np.median(marks))
print("Standard Deviation",np.std(marks))
```

```python
print("First Marks",marks[0])
print("Last number",marks[-1])
```

```python
print("Index 1 to 3:",marks[1:4])
print("First three ",marks[0:3])
print("Every second elements:",marks[::2])
```

```python
student_marks=np.array([[78,78,90,
                       [89,87,98,
                       [87,88,97]])
print("Student Marks",student_marks)
print("Shape",student_marks.shape)
```

```python
student_marks=np.array([[78,78,90],
                        [89,87,98],
                        [87,88,97]])
print("Student Marks",student_marks)
print("Shape",student_marks.shape)
```

```python
print("Mean of all values",student_marks.mean())
print("Mean of each column",student_marks.mean(axis=0))
print("Mean of each row",student_marks.mean(axis=1))
```
